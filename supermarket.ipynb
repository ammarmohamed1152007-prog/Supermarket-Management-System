# 🛒 Supermarket Management System (OOP Project)

## 📌 Project Description
You are required to design and implement a **Supermarket Management System** using **Python OOP principles**.  
The system should simulate the roles and operations inside a supermarket. Each major role must be implemented as a **separate class in a `.py` file**, and finally, you should integrate them all inside a **Jupyter Notebook (`.ipynb`)**.

The project should cover **multiple user types**:
- **Admin**: manages products and employees.  
- **Employee**: processes orders and assists customers.  
- **Customer**: shops for products, adds items to a cart, and makes purchases.  
- **Product**: represents an item with attributes (id, name, price, quantity).  
- **Inventory**: manages the list of products (add, remove, update, list).  
- **Order**: represents a customer order with items, total amount, and status.  
- **Payment**: handles customer payments and generates receipts.  

---

## 🎯 Requirements
1. Each class must be implemented in its own `.py` file.  
2. The system must allow interaction using a **menu-driven interface** (number-based menu for actions).  
3. Users should be able to log in as **Admin**, **Employee**, or **Customer**, and access features specific to their role.  
4. Demonstrate **Encapsulation, Inheritance, and Polymorphism** in your design.  
5. No external libraries should be used — only **pure Python OOP**.  
6. At the end, you must integrate everything inside a Jupyter Notebook (`.ipynb`) and run the system from there.

---

## 🧑‍💻 Features per Role

### 👑 Admin
- Add / Remove / Update products.  
- Hire / Fire employees.  
- View all products in inventory.  

### 👔 Employee
- View all products.  
- Process customer orders.  
- Update order status.  

### 🛍️ Customer
- View all available products.  
- Add products to cart.  
- Checkout and make payment.  
- Receive a receipt after purchase.  

### 📦 Inventory
- Store all products.  
- Provide methods to search, update, and display items.  

### 📑 Order
- Maintain list of products selected by the customer.  
- Calculate total amount.  
- Track order status (Pending, Processed, Completed).  

### 💳 Payment
- Accept payment from the customer.  
- Verify payment amount.  
- Generate a receipt.
